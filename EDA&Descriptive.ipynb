# Numerical Descriptive Statistics

```python
# Central Tendacy (Mean, Median, Mode)
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
print("Numeric columns:", numeric_cols)

central_tendency = {}

for col in numeric_cols:
    mode_val = df[col].mode()
    central_tendency[col] = {
        "mean": df[col].mean(),
        "median": df[col].median(),
        "mode": mode_val.iloc[0] if not
        mode_val.empty else None
    }
pd.DataFrame(central_tendency).T
```

```python
# Dispersion (Std, Variance, Range, IQR)

dispersion = {}

for col in numeric_cols:
    dispersion[col] = {
        "std": df[col].std(),
        "variance": df[col].var(),
        "min": df[col].min(),
        "max": df[col].max(),
        "range": df[col].max() - df[col].min(),
        "IQR": df[col].quantile(0.75) - df[col].quantile(0.25)
        
    }
    pd.DataFrame(dispersion).T
```

```python
#Top 12 cities by crime count

import seaborn as sns

df = pd.read_csv("crime_clean.csv")
df.columns = [c.strip() for c in df.columns]
if 'City' in df.columns:
    df = df.rename(columns={'City':'city'})
# get top
top_n = 12
top = df['city'].value_counts().head(top_n).reset_index()
top.columns = ['city','count']

# plot
plt.figure(figsize=(10,5))
ax = sns.barplot(data=top, y='city', x='count', orient='h')
plt.title(f"Top {top_n} Cities by Crime Count", fontsize=16)
plt.xlabel("Crime Count")
plt.ylabel("City")
# annotate values at end of bars
for p in ax.patches:
    ax.annotate(f'{int(p.get_width()):,}', (p.get_width()+50, p.get_y() + p.get_height()/2),
                va='center', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
#Crime share by city
city_crime = df['City'].value_counts().head(8)

plt.figure(figsize=(7,7))
plt.pie(city_crime.values, labels=city_crime.index, autopct='%1.1f%%')
plt.title("Crime Distribution by City (Top 8 Cities)")
plt.show()
```

```python
#Crime trend per year

# Convert date column
df['date'] = pd.to_datetime(df['Dispatch Date / Time'], errors='coerce')
df['year'] = df['date'].dt.year

# Select top 5 highest-crime cities
top_cities = df['city'].value_counts().head(5).index
trend_df = df[df['city'].isin(top_cities)]

# Group by year and city
trend = trend_df.groupby(['year','city']).size().unstack()

plt.figure(figsize=(6,3))
plt.plot(trend)
plt.title("Yearly Crime Trends of Top 5 Cities")
plt.xlabel("Year")
plt.ylabel("Crime Count")
plt.legend(title="City")
plt.tight_layout()
plt.show()
```

# RQ5.Where the geographic crime hotspots?

Theory link: Environmental criminology / hotspots theory-crime clusters in predictable high-activity zones

Author: Hasini Dilshari

```python
# HeatMap on a city Map

import pandas as pd
import folium
from folium.plugins import HeatMap
import warnings
warnings.filterwarnings('ignore')


df = pd.read_csv("crime_clean.csv")

# Create map centered on average coordinates
map_center = [df['Latitude'].mean(), df['Longitude'].mean()]
crime_map = folium.Map(location=map_center, zoom_start=6)

# Add heatmap layer
heat_data = df[['Latitude', 'Longitude']].dropna().values.tolist()
HeatMap(heat_data, radius=10).add_to(crime_map)

crime_map
```

```python
# Top Crime Hotspot Location

import matplotlib.pyplot as plt

top_areas = df['City'].value_counts().head(10)

plt.figure(figsize=(12,6))
plt.bar(top_areas.index, top_areas.values, color='green')
plt.xticks(rotation=45)
plt.title("Top 10 Crime Hotspot Cities")
plt.xlabel("City")
plt.ylabel("Crime Count")
plt.show()
```
