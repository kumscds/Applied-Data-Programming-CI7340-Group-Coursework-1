# Descriptive Statistics Full Section

```python
# load cleaned data set
df = pd.read_csv("crime_clean.csv")
df.head()
```

```python
# Standardise column names
df.columns = [c.strip().replace("/", "_").replace(" ", "_") for c in df.columns]
df.columns
```

```python
# Basic Data Summary
df.info()
df.describe(include="all")
```

# Numerical Descriptive Statistics

```python
# Central Tendacy (Mean, Median, Mode)
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
print("Numeric columns:", numeric_cols)

central_tendency = {}

for col in numeric_cols:
    mode_val = df[col].mode()
    central_tendency[col] = {
        "mean": df[col].mean(),
        "median": df[col].median(),
        "mode": mode_val.iloc[0] if not
        mode_val.empty else None
    }
pd.DataFrame(central_tendency).T
```

```python
# Dispersion (Std, Variance, Range, IQR)

dispersion = {}

for col in numeric_cols:
    dispersion[col] = {
        "std": df[col].std(),
        "variance": df[col].var(),
        "min": df[col].min(),
        "max": df[col].max(),
        "range": df[col].max() - df[col].min(),
        "IQR": df[col].quantile(0.75) - df[col].quantile(0.25)
        
    }
    pd.DataFrame(dispersion).T
```

# Categorical Descriptive Statistics
we detect column names automatically based on common patterns

```python
# Identify Key categorical columns

#Crime type
crime_col = [c for c in df.columns if "crime" in c.lower()][0]
# District
district_col = [c for c in df.columns if "district" in c.lower()][0]
#city
city_col = [c for c in df.columns if "city" in c.lower()][0]

crime_col, district_col, city_col
```

```python
# Crime Frequency table
crime_freq = pd.DataFrame({
    "count": df[crime_col].value_counts(),
    "percent": df[crime_col].value_counts(normalize=True) * 100
})
crime_freq.head(10)
```

```python
# Police District frequency
district_freq = pd.DataFrame({
    "count": df[district_col].value_counts(),
    "percent": df[district_col].value_counts(normalize=True) * 100
})
district_freq
```

```python
# City Distribution
city_freq = pd.DataFrame({
    "count": df[city_col].value_counts(),
    "percent": df[city_col].value_counts(normalize=True) * 100
})
city_freq
```

## Numerical Summaries - Latitude/Longitude/Time Fields

```python
#Identify Lat/Long Columns
lat_col = [c for c in df.columns if "lat" in c.lower()][0]
lon_col = [c for c in df.columns if "lon" in c.lower()][0]

lat_col, lon_col
```

```python
#Summary of Latitude and Longitude
df[[lat_col, lon_col]].describe()
```

```python
#Parse Time Field
time_cols = [c for c in df.columns if "time" in c.lower() or "date" in c.lower()]
time_cols
```

```python
#Assuming first datetime clumn

dt_col = time_cols[0]

df[dt_col] = pd.to_datetime(df[dt_col], errors='coerce')

df['Year'] = df[dt_col].dt.year
df['Month'] = df[dt_col].dt.month
df['Day'] = df[dt_col].dt.day
df['Hour'] = df[dt_col].dt.hour

df[['Year', 'Month', 'Day', 'Hour']].describe()
```

```python
#Full numerical summary using describe()
df.describe().T
```
